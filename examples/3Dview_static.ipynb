## Example of 3D display of .xlsx file
The main .xlsx file used for the positions and orientations of sensors, impacts, channels and virtual points can be simply visualized with the pyFBS. For the 3D visualization a python package pyvista is used.
#### 3D View
Open 3Dviewer in the background, with CSYS in origin

```python
view3D = pyFBS.display.view3D()
```

Add a structure from .stl file to the 3D 

```python
AB_stl = "../data/AM_substructuring_testbench/STL/AM_AB_final.stl"
view3D.add_stl(AB_stl)
```

#### Accelerometers
Add accelerometers from .xlsx file together with appropriate labels

```python
AB = "../data/AM_substructuring_testbench/Measurements/decoupling/Excel/AM_Measurements.xlsx"
df = pd.read_excel(AB, sheetname='Sensors_AB')

view3D.show_acc(df)
#view3D.label_acc(df)
```

#### Channels
Add corresponding channels from .xlsx file together with appropriate labels

```python
df = pd.read_excel(AB, sheetname='Channels_AB')

view3D.show_chn(df)
view3D.label_chn(df)
```

#### Impacts
Add impacts from .xlsx with appropriate labels

```python
df = pd.read_excel(AB, sheetname='Impacts_AB')

view3D.show_imp(df)
view3D.label_imp(df)
```

#### Virtual Points
Add unique VPs from .xlsx with appropriate labels

```python
df = pd.read_excel(AB, sheetname='VP_Channels')

view3D.show_vp(df)
view3D.label_vp(df)
```

#### Additional commands

The 3D viewer can be cleared by a simple command

```python
view3D.plot.clear()
```
