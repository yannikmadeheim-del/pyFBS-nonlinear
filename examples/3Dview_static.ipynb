## Example of 3D display of .xlsx file
The main .xlsx file used for the positions and orientations of sensors, impacts, channels and virtual points can be simply visualized with the pyFBS. For the 3D visualization a python package pyvista is used.
#### 3D Viewer
Open 3Dviewer in the background, with CSYS in origin

```python
a = view3D()
```

Add a structure from .stl file to the 3D 

```python
AB = "../data/AM_substructuring_testbench/STL/AM_AB_final.stl"
a.add_stl(AB)
```

#### Accelerometers
Add accelerometers from .xlsx file together with appropriate labels

```python
AB = "../data/AM_substructuring_testbench/Measurements/decoupling/Excel/AM_Measurements.xlsx"
df = pd.read_excel(AB, sheetname='Sensors_AB')

a.show_acc(df)
a.label_acc(df)
```

#### Channels
